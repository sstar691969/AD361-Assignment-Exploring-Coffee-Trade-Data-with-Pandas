```python

# Part 3: Calculate Summary Statistics for Total Values

# Create a list containing each dataset, its name, and its total-value column.
for name, df, total_column in [
    ("Coffee Exports", coffee_export, "Total_export"),
    ("Coffee Imports", coffee_import, "Total_import"),
    ("Coffee Production", coffee_production, "Total_production")
]:
    # Print the dataset name and the column being analyzed.
    print(f"\n{name} — {total_column}")

    # Calculate the minimum, maximum, and average (mean) total values.
    # The agg() method applies all three calculations to the selected column.
    # The round(2) method rounds the results to two decimal places.
    print(df[total_column].agg(["min", "max", "mean"]).round(2))
```
